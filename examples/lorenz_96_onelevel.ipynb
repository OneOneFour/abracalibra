## lorenz 96 one level!

Calibrate using variable $F$ from uniform forcing all the way up to N  (1,2,3,9,12,18,36) parameter variants!

```python
import torch
import matplotlib.pyplot as plt 
from lorenz import Lorenz96OneLevelScaled

T = 50
dt =0.005
F = 1
adv = 1
diff = 1
N = 36 


x0 = torch.ones(100,N) + torch.randn(100,N)*0.1 
model = Lorenz96OneLevelScaled(F,adv,diff)
r = model(T,dt,x0)
```

```python
t = torch.arange(0,T,dt)
n = torch.arange(0,N).float()
Ns,Ts = torch.meshgrid(n,t,indexing='xy')

plt.contourf(Ns,Ts,r[0]) 
```
